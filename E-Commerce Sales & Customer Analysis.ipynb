```python
## 6. Product Analysis
```

```python
product_revenue = df.groupby("Description")["Revenue"].sum().sort_values(ascending=False)
```

```python
## 7. Country & Monthly Revenue Analysis
```

```python
country_revenue = df.groupby("Country")["Revenue"].sum().sort_values(ascending=False)
country_revenue.head(10)
```

```python
monthly_revenue = df.groupby(df["InvoiceDate"].dt.to_period("M"))["Revenue"].sum()

monthly_revenue
```

```python
customer_orders.head(10)
```

```python
customer_summary = df.groupby("CustomerID").agg(
    Orders=("InvoiceNo", "nunique"),
    Revenue=("Revenue", "sum")
)

customer_summary.sort_values("Revenue", ascending=False).head(10)
```

```python
country_aov = df.groupby("Country")["Revenue"].sum() / df.groupby("Country")["InvoiceNo"].nunique()
country_aov.sort_values(ascending=False).head(10)
```

```python
country_revenue.head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Countries by Revenue")
plt.xlabel("Country")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
product_revenue.head(10).plot(kind="bar", figsize=(10,5))

plt.title("Top 10 Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
## 8. SQL-Based Business Analysis
```

```python
import sqlite3
```

```python
conn = sqlite3.connect("ecommerce.db")

df.to_sql("ecommerce", conn, if_exists="replace", index=False)
```

```python
query = """
SELECT CustomerID, 
       ROUND(SUM(Revenue), 2) AS TotalRevenue
FROM ecommerce
GROUP BY CustomerID
ORDER BY TotalRevenue DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Country,
       ROUND(SUM(Revenue), 2) AS TotalRevenue
FROM ecommerce
GROUP BY Country
ORDER BY TotalRevenue DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT Description,
       ROUND(SUM(Revenue), 2) AS TotalRevenue
FROM ecommerce
GROUP BY Description
ORDER BY TotalRevenue DESC
LIMIT 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
SELECT strftime('%Y-%m', InvoiceDate) AS Month,
       ROUND(SUM(Revenue), 2) AS TotalRevenue
FROM ecommerce
GROUP BY Month
ORDER BY Month;
"""

pd.read_sql_query(query, conn)
```
